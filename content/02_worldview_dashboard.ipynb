## Architectural contrast

The original WorldView is a React/TypeScript/Cesium application with an Express proxy and multiple external data integrations. This teaching version uses Leaflet, public browser-accessible endpoints, and JupyterLite so it can be hosted entirely as static files on GitHub Pages. That constraint is useful for learning: it makes data access, CORS, API-key handling, provenance, and graceful degradation explicit.
